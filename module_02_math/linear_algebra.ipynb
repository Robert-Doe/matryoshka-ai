# Module 02 — Linear Algebra You Can Feel

Linear algebra is the language models think in. You don't need proofs — you
need intuition for four objects: **scalars, vectors, matrices, and the dot
product**. Run each cell, change numbers, re-run. That's how it sinks in.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. A vector is a point AND a direction
`[3, 2]` is "go 3 right, 2 up". Length (norm) = how far from origin.

```python
v = np.array([3, 2])
print("vector:", v)
print("length (norm):", np.linalg.norm(v))

plt.figure(figsize=(4, 4))
plt.quiver(0, 0, v[0], v[1], angles="xy", scale_units="xy", scale=1, color="C0")
plt.xlim(-1, 5); plt.ylim(-1, 5); plt.grid(True)
plt.title("the vector [3, 2]")
plt.gca().set_aspect("equal")
plt.show()
```

## 2. The dot product = alignment = similarity
`a·b = |a||b|cos(angle)`. Same direction → large. Perpendicular → 0.
This is THE operation behind search, recommendations, and attention.

```python
a = np.array([2, 2])
b = np.array([3, 1])
c = np.array([-2, 2])   # perpendicular-ish to a
for name, w in [("b", b), ("c", c)]:
    cos = np.dot(a, w) / (np.linalg.norm(a) * np.linalg.norm(w))
    print(f"cosine(a, {name}) = {cos:+.3f}")
print("\n-> higher cosine means 'more similar direction'. Anomaly detection lives here.")
```

## 3. A matrix transforms space
Multiply a matrix by every point and the whole space stretches/rotates/flips.
A neural-network layer is exactly this: `output = W @ input + b`.

```python
# a rotation matrix (45 degrees)
theta = np.pi / 4
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
square = np.array([[0, 1, 1, 0, 0],
                   [0, 0, 1, 1, 0]])
rotated = R @ square

plt.figure(figsize=(4, 4))
plt.plot(square[0], square[1], "o-", label="original")
plt.plot(rotated[0], rotated[1], "o-", label="rotated 45°")
plt.legend(); plt.grid(True); plt.gca().set_aspect("equal")
plt.title("a matrix rotates the square")
plt.show()
```

## 4. Matrix multiply = a whole layer at once
One `@` transforms an entire batch of feature vectors. This is why GPUs
(which do giant matrix multiplies fast) power deep learning.

```python
batch = np.random.rand(4, 3)          # 4 examples, 3 features each
W = np.random.rand(3, 2)              # weights: 3 features -> 2 outputs
b = np.array([0.1, -0.1])
out = batch @ W + b                   # the core of a neural layer
print("input batch shape :", batch.shape)
print("weights shape     :", W.shape)
print("output shape      :", out.shape, "  (4 examples, 2 outputs)")
```
